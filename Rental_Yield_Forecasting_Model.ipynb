**Strategic Objectives**
**Capital Allocation Accuracy:** Evaluate investment opportunities using realistic Net Operating Income (NOI) after accounting for full operational expenditure (OpEx) and vacancy allowances.

**Client Value Transparency:** Provide prospective buyers and investors with clear, institutional-grade yield forecasts for luxury cluster developments (e.g., Avondale assets).

**Operational Expense Benchmarking**: Standardize the evaluation of complex HOA levies (covering 24/7 security, backup power, borehole maintenance, and shared amenities) against rental returns.

**Business & Portfolio Impact**
Accelerated Sales Conversion: High-yield visual insights enable private wealth and corporate clients to make faster, data-backed acquisition decisions.

**Risk Mitigation:** Factoring vacancy risk and interior maintenance reserves directly into underwriting protects investor return expectations and reduces portfolio drawdown risk.

```python
import pandas as pd

def forecast_rental_yield(
    purchase_price: float,
    gross_monthly_rent: float,
    monthly_hoa_levies: float,
    management_fee_pct: float = 0.08,
    annual_vacancy_rate: float = 0.05,
    annual_maintenance_pct: float = 0.02
) -> dict:
    """
    Calculates Gross and Net Rental Yields for luxury residential clusters
    managed or targeted by Jacobs Real Estate.
    """
    # Gross Income Analysis
    gross_annual_rent = gross_monthly_rent * 12
    effective_gross_income = gross_annual_rent * (1 - annual_vacancy_rate)

    # Operating Expense Breakdown (OpEx)
    annual_hoa = monthly_hoa_levies * 12
    annual_management = effective_gross_income * management_fee_pct
    annual_maintenance = purchase_price * annual_maintenance_pct
    total_operating_expenses = annual_hoa + annual_management + annual_maintenance

    # Net Income & Yield Calculations
    net_operating_income = effective_gross_income - total_operating_expenses
    gross_yield = (gross_annual_rent / purchase_price) * 100
    net_yield = (net_operating_income / purchase_price) * 100

    return {
        "Brokerage / Entity": "Jacobs Real Estate",
        "Purchase Price": f"${purchase_price:,.2f}",
        "Gross Annual Rent": f"${gross_annual_rent:,.2f}",
        "Effective Gross Income (adj. vacancy)": f"${effective_gross_income:,.2f}",
        "Total Annual OpEx (HOA + Mgmt + Maint)": f"${total_operating_expenses:,.2f}",
        "Net Operating Income (NOI)": f"${net_operating_income:,.2f}",
        "Gross Rental Yield (%)": f"{gross_yield:.2f}%",
        "Net Rental Yield (%)": f"{net_yield:.2f}%"
    }

# --- Case Study: 4-Bed Luxury Cluster Unit (The Jacobs, Avondale) ---
jacobs_forecast = forecast_rental_yield(
    purchase_price=350000,          # Market acquisition price
    gross_monthly_rent=2400,        # Projected luxury monthly rent
    monthly_hoa_levies=250,         # 24/7 Security, borehole, solar/generator, pool & gym
    management_fee_pct=0.08,        # Jacobs Real Estate standard agency fee (8%)
    annual_vacancy_rate=0.05,       # 5% conservative vacancy reserve
    annual_maintenance_pct=0.015    # 1.5% interior maintenance reserve
)

print("=" * 50)
print(" JACOBS REAL ESTATE — INVESTMENT YIELD ANALYSIS")
print("=" * 50)
for key, val in jacobs_forecast.items():
    print(f"{key:<40} : {val}")
```
